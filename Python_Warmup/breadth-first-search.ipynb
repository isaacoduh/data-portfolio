## Breadfirst Search

The search algorithm is an algorithm to retrieve information stored within some data structure, or calculated in the search space of a problem domain. Breadth-first search is an algorithm for transversing or searching a tree or graph data structures. It starts at the root node and explores all nodes at the present depth before moving forward on the nodes at the next depth level. \n

In other words, it expands the shallowest unexpanded node which can be implemented by a First-In-First-Out (FIFO) queue.

From the above step by step expansion diagram, we can see that the BFS algorithm prioritizes edges closest to the starting vertex to perform a search.

Now, let's evaluate this algorithm:
Denote d as the depth of the least-cost solution and b as the maxiumum branching factor for teh search tree or graph. Assume a hypothetical state space where every node can be expanded into b new nodes, solution of path length d:


1. Time Complexity: How long does it take to find a solution?
    1 + b + b^2 + b^3 + ... + b^d = O(b^d)
2. Space Complexity: Maximum number of nodes in memory
    Keeps every node in memory = O(b^d)
3. Completenes: Does it find a solution if one exists?
    yes
4. Optimality: Does it always find the best (least-cost) solution?
    Yes, when all steps cost equally.

```python
# Creating the function that takes edges of the graph which outputs the adjacencylist for undirected graphs

from collections import defaultdict
# generate adjacency list 
def generate_adjacency_list(edges):
    adjacencyList = defaultdict(list)
    for u, v in edges:
        adjacencyList[u].append(v)
        adjacencyList[v].append(u)
    return adjacencyList

edges = [['A', 'B'], ['A', 'C'], ['B', 'D'], ['B', 'E'], ['C', 'F'], ['C', 'G'], ['D', 'H'], ['D', 'I'], ['E', 'J'], ['E', 'K'], ['F', 'L'], ['F', 'M']]
adjacencyList = generate_adjacency_list(edges)
print(adjacencyList)
```

```python
# Function that takes the adjacency list and starting vertex which output the sequence of BFS search.

# find bfs traversal from starting vertex
def bfs(adjacencyList, vertex):
    visitedSet = set()
    queue = []
    visitedSet.add(vertex)
    queue.append(vertex)

    result = []
    print(visitedSet)
    print(queue)

    while queue:
        v = queue[0]
        result.append(v)
        print(result)
        queue = queue[1:]
        print(queue)
        for neighbor in adjacencyList[v]:
            print(neighbor)
            print(adjacencyList[v])
            if neighbor not in visitedSet:
                print(visitedSet)
                visitedSet.add(neighbor)
                print(visitedSet)
                queue.append(neighbor)
                print(queue)
    return result

print(bfs(adjacencyList, 'A'))
```

```python
payroll2021_df.rename(columns={'AgencyCode': 'AgencyID'}, inplace=True)
```

```python
# convert string columns to numerical values
numeric_columns = ['BaseSalary', 'RegularHours', 'RegularGrossPaid', 'OTHours', 'TotalOTPaid', 'TotalOtherPay']
```

```python
# Convert via a forloop 
for col in numeric_columns:
    if col in payroll2020_df.columns:
        payroll2020_df[col] = pd.to_numeric(payroll2020_df[col], errors="coerce")
    if col in payroll2021_df.columns:
        payroll2021_df[col] = pd.to_numeric(payroll2021_df[col], errors="coerce")
```

```python
# Confirm that columns have been affected
payroll2020_df.head()
```

```python
# change to datetime for agency start date colum
if "AgencyStartDate" in payroll2020_df.columns:
    payroll2020_df['AgencyStartDate'] = pd.to_datetime(payroll2020_df['AgencyStartDate'], errors='coerce')

if "AgencyStartDate" in payroll2021_df.columns:
    payroll2021_df['AgencyStartDate'] = pd.to_datetime(payroll2021_df['AgencyStartDate'], errors="coerce")
```
